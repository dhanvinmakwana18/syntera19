# Syntera v2.0 - Programmatic Quickstart
Welcome to the **Syntera Enterprise Data Scientist** notebook! This tutorial shows how to use the underlying API and Agents programmatically in your own pipelines.

```python
import sys
import os
from pathlib import Path

# Add root directory to path
sys.path.append(str(Path.cwd().parent))

from backend.agents.data_scientist import DataScientistAgent
import pandas as pd
```

## 1. Load Example Data (Titanic)

```python
# Create a mock CSV for demonstration
df = pd.DataFrame({
    'PassengerId': range(1, 11),
    'Survived': [0, 1, 1, 1, 0, 0, 0, 1, 1, 1],
    'Pclass': [3, 1, 3, 1, 3, 3, 1, 3, 3, 2],
    'Age': [22, 38, 26, 35, 35, None, 54, 2, 27, 14],
    'Fare': [7.25, 71.28, 7.92, 53.1, 8.05, 8.45, 51.86, 21.07, 11.13, 30.07]
})
df.to_csv('titanic_subset.csv', index=False)
df.head()
```

## 2. Initialize and Run the Agent

```python
# Set your API Key if not already in .env
# os.environ['NVIDIA_API_KEY'] = 'your_key_here'

agent = DataScientistAgent()
print("Agent Initialized! Starting analysis...")

result = agent.analyze(['titanic_subset.csv'], goal="Predict survival based on Age, Pclass, and Fare.")
print("\nData Quality Score:", result['eda_summary']['data_quality_score'])
```

## 3. View the Results & PDF Report

```python
print("Best Recommended Model:", result['best_model']['name'])
print("Accuracy:", result['best_model']['accuracy'])

pdf_path = result['report_pdf']
print(f"\nFull Professional PDF Report saved to: {pdf_path}")
```
